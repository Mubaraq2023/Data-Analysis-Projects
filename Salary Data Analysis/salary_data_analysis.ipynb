## 5. Outlier Detection and Removal

### Visualize Salary Outliers

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x = df["Salary"])

plt.title("Salary Outliers")
plt.show()
```

### Interquartile Range

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
```

```python
outliers = df[(df["Salary"] < lower_limit) | (df["Salary"] > upper_limit)]
outliers
```

```python
# df["Salary"] = df["Salary"].clip(lower = lower_limit, upper = upper_limit)
# df
```

```python
df = df[(df["Salary"] >= lower_limit) & (df["Salary"] <= upper_limit)]
df
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(x = df["Salary"])

plt.title("Salary Outliers")
plt.show()
```

## 7. Answering Business Questions

### Q1: Which Job Roles Have the Highest Average Salary?

```python
df["Job Roles"].nunique()
```

```python
df["Job Roles"].unique()
```

```python
jobrole_avg_salary = df.groupby("Job Roles")["Salary"].mean().sort_values(ascending=False).head(10)
```

```python
jobrole_avg_salary
```

```python
plt.figure(figsize=(8, 5))
jobrole_avg_salary.plot(kind="bar")
plt.title("Top 10 Job Roles by Average Salary")
plt.ylabel("Average Salary")
plt.show()
```

Database roles have the highest average salary at around ₹7 lakh, followed closely by SDE, iOS and Mobile roles above ₹5.8 lakh. Python and Testing roles have the lowest average pay among the top 10 but the gap between the highest and lowest is not very large (around ₹2.5 lakh).

### Q2: Which Cities Offer the Highest Average Salary

```python
cities_avg_sal = df.groupby("Location")["Salary"].mean().sort_values(ascending=False)
```

```python
cities_avg_sal
```

```python
plt.figure(figsize=(8, 5))
colors = sns.color_palette("viridis", len(cities_avg_sal))
cities_avg_sal.plot(kind="bar", color=colors, edgecolor="black")

plt.title("Top Cities by Average Salary")
plt.ylabel("Average Salary")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

Mumbai has the highest average salary among all cities at close to ₹7 lakh, noticeably ahead of the rest. The other cities are fairly close to each other, ranging between ₹5 lakh and ₹5.7 lakh, showing relatively even pay across most locations except Mumbai.

### Q6: Check and Show the relationship between Salaries and Ratings

```python
df.corr(numeric_only=True)
```

```python
df["Rating"].corr(df["Salary"])
```

```python
sns.relplot(x = "Rating", y = "Salary", data=df, alpha=0.5)

plt.title("Rating vs Salary", fontweight="bold")
plt.show()
```

There is a weak positive link between rating and salary, higher-rated employees tend to have a slightly wider range of high salaries, especially between ratings 3 and 5. However, the spread is very wide at every rating level, showing that rating alone does not strongly determine salary

```python
sns.regplot(x = "Rating", y = "Salary", data=df, scatter_kws={"alpha":0.5, "color":"teal"}, line_kws={"linewidth":3})

plt.title("Rating vs Salary (with trend line)", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.barplot(x = "Employment Status", y = "Salary", data=df)
plt.title("Salary by Employment Status", fontweight="bold")
plt.ticklabel_format(style="plain", axis="y")
plt.xlabel("Employment Type")
plt.ylabel("Salary")
plt.show()
```

Full Time employees have the highest average salary (close to ₹5.8 lakh) with a small error range, showing consistent pay across this group. Trainees have the lowest average salary and the largest error bar, meaning their pay varies a lot from person to person.

### Q8: Which roles are most common?

```python
top_job_roles = df["Job Roles"].value_counts().sort_values(ascending=True)
top_job_roles
```

```python
plt.figure(figsize=(10,6))
colors = sns.color_palette("hsv", len(top_job_roles))
plt.barh(top_job_roles.index, top_job_roles.values, color = colors, edgecolor="black")

for i, v in enumerate(top_job_roles.values):
    plt.text(v + 50, i, str(v), va = "center", fontweight = "bold")
    
plt.title("Top 10 most common Job Roles", fontsize=16, fontweight="bold")
plt.xlabel("No of Records", fontsize=12)
plt.ylabel("Job Roles", fontsize=12)

plt.show()
```

SDE is by far the most common job role, with 7,245 records more than double the next most common role, Android (2,812). Mobile is the least common role in the top 10 with only 211 records, showing a big gap between the most and least represented roles.

### Q9: How does average salary change as company rating increases?

```python
rating_salary = df.groupby("Rating")["Salary"].mean().reset_index()
rating_salary.head()
```

```python
plt.figure(figsize=(7, 5))
sns.lineplot(x = "Rating", y = "Salary", data = rating_salary, marker = "o")

plt.title("Average Salary trend across Company Ratings", fontsize = 14, fontweight = "bold")
plt.xlabel("Company Rating")
plt.ylabel("Average Salary")
plt.tight_layout()

plt.show()
```

Overall, average salary tends to rise as company rating increases, moving from around ₹4 lakh at rating 1 to over ₹5.5 lakh at rating 5. There is a sharp, unusual spike near rating 1.7 and some dips (like near rating 4), showing the trend is not perfectly smooth and has a few outlier points.

## Mini Dashboard

```python
# Prepare data
rating_salary = df.groupby("Rating")["Salary"].mean().reset_index()
top_locations = df.groupby("Location")["Salary"].mean().sort_values(ascending = False)

# Create Dashboard Layout
plt.figure(figsize=(14, 10))

#1. Salary Distribution
plt.subplot(2,2,1)
sns.histplot(df["Salary"], bins = 30, kde = True)
plt.title("Salary Distribution")

#2. Rating vs Salary
plt.subplot(2,2,2)
sns.scatterplot(x = "Rating", y = "Salary", data=df, alpha=0.5)
sns.regplot(x = "Rating", y = "Salary", data=df, scatter = False, color = "red")
plt.title("rating vs Salary")

#3. Average Salary by Rating
plt.subplot(2,2,3)
sns.lineplot(x = "Rating", y = "Salary", data = rating_salary, marker = "o")
plt.title("Average Salary by Rating")

#4. Top Locations
plt.subplot(2,2,4)
top_locations.sort_values().plot(kind="barh", color = "green")
plt.title("Top 10 Locations by Average Salary")

plt.suptitle("Salary Data Analysis Dashboard", fontsize=18, fontweight= "bold")

plt.tight_layout()
plt.show()
```
